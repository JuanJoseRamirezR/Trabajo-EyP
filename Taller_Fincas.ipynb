```python
import pandas as pd

# Ejercicio 1: Lectura y Verificación Estructural
# Cargamos el archivo asegurando que no se generen columnas de índices numéricas duplicadas
df_fincas = pd.read_csv('fincas_produccion_cafe.csv', index_col=False)

# Mostramos las primeras 5 filas para verificar la carga
df_fincas.head()
```

```python
# Ejercicio 2: Selección Unidimensional y Proyección Multivariable
# Extraemos la columna Variedad como una Serie
serie_variedad = df_fincas['Variedad']

# Obtenemos una vista de tabla con Municipio, Area_Hectareas y Kilos_Recolectados
vista_multivariable = df_fincas[['Municipio', 'Area_Hectareas', 'Kilos_Recolectados']]

# Mostramos las primeras 5 filas de la vista
vista_multivariable.head()
```

```python
# Ejercicio 3: Acceso por Etiqueta (loc)
# Consulta todos los datos del predio situado en la posición de índice 280
predio_280 = df_fincas.loc[280]
print("--- Datos del predio 280 ---")
print(predio_280)

# Ejercicio 4: Eliminación de Filas Duplicadas
# Aplicamos .drop_duplicates() para mantener la base limpia
df_fincas = df_fincas.drop_duplicates()
print("\n--- Limpieza de duplicados ---")
print(f"Total de filas únicas restantes: {len(df_fincas)}")
```

```python
# Ejercicio 6: Filtrado Categórico por Lista (isin)
# Seleccionamos predios cultivados en Anserma y Risaralda
fincas_anserma_risaralda = df_fincas[df_fincas['Municipio'].isin(['Anserma', 'Risaralda'])]

# Mostramos el resultado
print(f"Fincas en Anserma y Risaralda: {len(fincas_anserma_risaralda)}")
fincas_anserma_risaralda.head()
```

```python
# Ejercicio 7: Filtrado por Intervalo Numérico (between)
# Aislamos los lotes cuya extensión esté entre 3.0 y 8.0 hectáreas inclusive
lotes_medianos = df_fincas[df_fincas['Area_Hectareas'].between(3.0, 8.0)]

print(f"Lotes de tamaño mediano (3 a 8 ha): {len(lotes_medianos)}")
lotes_medianos.head()
```

```python
# Ejercicio 8: Consultas Directas con .query()
# Extraemos fincas con certificación Orgánica y más de 5000 kilos recolectados
organicos_alta_produccion = df_fincas.query("Certificacion == 'Orgánico' and Kilos_Recolectados > 5000")

print(f"Fincas orgánicas de alta producción: {len(organicos_alta_produccion)}")
organicos_alta_produccion.head()
```

```python
# Ejercicio 9: Filtrado Compuesto con Métodos de Texto
# Construimos una condición lógica múltiple
condicion_lavado = df_fincas['Metodo_Beneficio'].str.startswith('Lavado', na=False)
condicion_geisha = df_fincas['Variedad'] == 'Geisha'

# Aplicamos ambas condiciones con el operador &
lavado_geisha = df_fincas[condicion_lavado & condicion_geisha]

print(f"Fincas de variedad Geisha con método Lavado: {len(lavado_geisha)}")
lavado_geisha.head()
```

```python
# Ejercicio 10: Imputación de Valores Faltantes
# Calculamos el promedio del área
promedio_area = df_fincas['Area_Hectareas'].mean()

# Rellenamos los vacíos (NaN)
df_fincas['Area_Hectareas'] = df_fincas['Area_Hectareas'].fillna(promedio_area)
df_fincas['Kilos_Recolectados'] = df_fincas['Kilos_Recolectados'].fillna(0.0)

# Verificamos que ya no hay valores nulos en esas columnas
print("Valores vacíos restantes:")
print(df_fincas[['Area_Hectareas', 'Kilos_Recolectados']].isnull().sum())
```

```python
# Ejercicio 11: Transformación y Creación de Columnas con .apply()
# Definimos la función matemática
def calcular_rendimiento(fila):
    if fila['Area_Hectareas'] == 0:  # Prevenir división por cero
        return 0
    return fila['Kilos_Recolectados'] / fila['Area_Hectareas']

# Aplicamos la función para crear la nueva columna
df_fincas['Rendimiento_Kilos_Ha'] = df_fincas.apply(calcular_rendimiento, axis=1)

# Mostramos el resultado de la nueva columna
df_fincas[['ID_Finca', 'Area_Hectareas', 'Kilos_Recolectados', 'Rendimiento_Kilos_Ha']].head()
```

```python
# Ejercicio 12: Renombrado y Reordenamiento de Atributos
# Renombramos usando un diccionario
df_fincas = df_fincas.rename(columns={'Metodo_Beneficio': 'Beneficio', 'Altitud_MSNM': 'Altitud'})

# Reorganizamos el orden de las columnas manualmente
columnas_ordenadas = ['ID_Finca', 'Rendimiento_Kilos_Ha', 'Municipio', 'Variedad', 
                      'Beneficio', 'Area_Hectareas', 'Kilos_Recolectados', 'Altitud', 'Certificacion']
df_fincas = df_fincas[columnas_ordenadas]

df_fincas.head()
```

```python
# Ejercicio 13: Agrupación y Métricas de Producción por Municipio
metricas_municipio = df_fincas.groupby('Municipio').agg({
    'Altitud': 'mean',
    'Kilos_Recolectados': 'sum'
})
print("--- MÉTRICAS POR MUNICIPIO ---")
print(metricas_municipio, "\n")

# Ejercicio 14: Dispersión con Función Personalizada (rango = max - min)
def rango(series):
    return series.max() - series.min()

dispersion_variedad = df_fincas.groupby('Variedad')[['Area_Hectareas', 'Kilos_Recolectados']].agg(rango)
print("--- DISPERSIÓN POR VARIEDAD (Rango) ---")
print(dispersion_variedad, "\n")

# Ejercicio 15: Agrupamiento Multinivel (Por Municipio y Variedad a la vez)
multinivel = df_fincas.groupby(['Municipio', 'Variedad']).agg({
    'Altitud': 'mean',
    'Kilos_Recolectados': 'sum'
})
print("--- AGRUPAMIENTO MULTINIVEL ---")
multinivel.head(10) # Mostramos solo las primeras 10 combinaciones
```
